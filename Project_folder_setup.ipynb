```r
# Clean the total_energy_emissions data
# Read the CSV
library(tidyverse)
df_raw <- read_csv("raw_data/total_energy_emissions.csv")

# Rename the first column to 'State' and filter for WA row only
df_wa_total_energy_emissions <- df_raw %>%
  rename(State = 1) %>%
  filter(State == "WA")

head(df_wa_total_energy_emissions)
```

```r
# Clean the per_capita_emissions data
# Read the CSV
library(tidyverse)
df_raw <- read_csv("raw_data/per_capita_emissions.csv")

# Rename the first column to 'State' and filter for WA row only
df_wa_per_capita <- df_raw %>%
  rename(State = 1) %>%
  filter(State == "WA")

head(df_wa_per_capita)
```

```r
# Write df_wa_per_capita to clean_data
write_csv(df_wa_per_capita, "clean_data/df_wa_per_capita.csv")

# Verify the file exists
list.files("clean_data/")
```

```r
# Write clean total_energy_emissions to clean_data
write_csv(df_wa_total_energy_emissions, "clean_data/total_energy_emissions_clean.csv")

# Verify the file exists
list.files("clean_data/")
```
