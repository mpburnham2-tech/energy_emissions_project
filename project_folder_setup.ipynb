```r
# Clean the total_energy_consumption_by_source data
library(tidyverse)

# Reload raw dataset
df_raw <- read_csv("raw_data/total_energy_consumption_by_source.csv")

# Filter for only years 1960 through 2020
# Pivot the dataset so that years (1960-2020) become columns in chronological order
df_wa_energy_consumption_by_source <- df_raw %>%
  filter(Year >= 1960 & Year <= 2020) %>%
  pivot_longer(cols = -Year, names_to = "Source", values_to = "Consumption") %>%
  pivot_wider(names_from = Year, values_from = Consumption) %>%
  select(Source, `1960`:`2020`)

head(df_wa_energy_consumption_by_source)
```

```r
# Clean the total_energy_emissions data
library(tidyverse)

# Reload total_energy_emissions dataset
df_raw <- read_csv("raw_data/total_energy_emissions.csv")

# Filter for only WA and years 1960 through 2020
df_wa_total_energy_emissions <- df_raw %>%
  filter(State == "WA") %>%
  select(State, `1960`:`2020`)

head(df_wa_total_energy_emissions)
```

```r
# Clean the wind_energy_consumption data
library(tidyverse)

# Reload wind_energy_consumption dataset from raw_data folder
df_raw <- read_csv("raw_data/wind_energy_consumption.csv")

# Filter for only WA and years 2002 through 2020
df_wa_wind_energy_consumption <- df_raw %>%
  filter(State == "WA") %>%
  select(State, `2002`:`2020`)

head(df_wa_wind_energy_consumption)
```

```r
# Clean the hydropower_consumption data
library(tidyverse)

# Reload the hydropower_consumption dataset from raw_data folder
df_raw <- read_csv("raw_data/hydropower_consumption.csv")

# Filter for only WA and years 2002 through 2020
df_wa_hydropower_consumption <- df_raw %>%
  filter(State == "WA") %>%
  select(State, `2002`:`2020`)

head(df_wa_hydropower_consumption)
```

```r
# Clean the solar_energy_consumption data
library(tidyverse)

# Reload solar_energy_consumption dataset from raw_data folder
df_raw <- read_csv("raw_data/solar_energy_consumption.csv")

# Filter for only WA and years 2002 through 2020
df_wa_solar_energy_consumption <- df_raw %>%
  filter(State == "WA") %>%
  select(State, `2002`:`2020`)

head(df_wa_solar_energy_consumption)
```

```r
# Clean the per_capita_emissions data
library(tidyverse)

# Reload the raw per-capita_emissions dataset
df_raw <- read_csv("raw_data/per_capita_emissions.csv")

# Filter for only WA and years 1960 through 2020
df_wa_per_capita_emissions <- df_raw %>%
  filter(State == "WA") %>%
  select(State, `1960`:`2020`)

head(df_wa_per_capita_emissions)
```
